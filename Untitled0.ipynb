```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

#deep modules
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.callbacks import EarlyStopping
```

Reading file

```python
df = pd.read_csv("housing.csv")
df.head()
```

Prepocessing

```python
df.shape
```

```python
df.info()
```

```python
df['ocean_proximity'].value_counts()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

Encoding

```python
df['ocean_proximity'].value_counts()
```

```python
df["ocean_proximity"]=df['ocean_proximity'].map({"<1H OCEAN":0,"INLAND":1,"NEAR OCEAN":2,"NEAR BAY":3,"ISLAND":4})
```

```python
df.head()
```

Train-test split

```python
x=df.drop('median_house_value', axis=1)
y=df["median_house_value"]
```

```python
x=np.array(x)
y=np.array(y)
```

```python
x
```

```python
x_train, x_test, y_train,  y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Scaling

```python
min_max_scaler=MinMaxScaler()
x_train=min_max_scaler.fit_transform(x_train)
x_test=min_max_scaler.transform(x_test)
```

```python
x_train
```

Model training and architecture

```python
x_train.shape
```

```python
model=Sequential([
    #input layer
    #Relu is mostly used in input and hidden layer, others are sigmoid, tangent is surpassed by relu in these operations
    Dense(1000,input_shape=[x_train.shape[1],],activation='relu'),
    #Hidden layer1
    Dense(500, activation='relu'),
    #Hidden layer2
    Dense(250,activation='relu'),
    #output layer
    Dense(1,activation='linear')
])

model.summary()
```

```python
model.compile(optimizer='rmsprop',loss='mse',metrics=['mae'])
es=EarlyStopping(monitor='val_loss',mode='min',patience=50,restore_best_weights=True)
```

```python
history=model.fit(x_train,y_train,validation_data=(x_test, y_test), callbacks=[es],epochs=10,batch_size=50,verbose=1)
```

input layer->hidden layer ->output layer==> forward propagation

Initialize weight, summation, activation, error

output layer->hidden layer ->input layer==> backward propagation

update weight, summation, activation, error


new_weight=old_weight-partial derivative of old error i.e.W1=W0-a(del(error)/W0)

regression project ma relu use hunchha, classification project ma sigmoid use hunchha

**Predictive System**

```python
def pred(longitude,latitude,housing_median_age,total_rooms,total_bedrooms,population,households,median_income,ocean_proximity):
  features=np.array([longitude,latitude,housing_median_age,total_rooms,total_bedrooms,population,households,median_income,ocean_proximity])
  feature_scaled=min_max_scaler.fit_transform([features])
  result=model.predict(feature_scaled).reshape(-1,1)
  return result[0]
```
